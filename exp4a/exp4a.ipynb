```python
import pandas as pd
df = pd.read_csv("insurance.csv")
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_charges = df[df["charges"] > 20000]
print(high_charges)
```

```python
smokers = df[df["smoker"] == "yes"]
print(smokers)
```

```python
sorted_df = df.sort_values(
    by="charges",
    ascending=False
)
print(sorted_df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
sorted_df = df.sort_values(by="bmi", ascending=False)
plt.figure(figsize=(12,5))
plt.bar(range(len(sorted_df)), sorted_df["bmi"])
plt.title("BMI Ranking")
plt.xlabel("Individuals (Sorted)")
plt.ylabel("BMI")
plt.show()
```
