```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("insurance.csv")

data = df.head()

plt.plot(data["age"], data["charges"])

plt.title("Age vs Medical Charges")
plt.xlabel("Age")
plt.ylabel("Medical Charges")

plt.show()
```

```python
plt.bar(data["age"], data["charges"])

plt.title("Age vs Medical Charges")
plt.xlabel("Age")
plt.ylabel("Medical Charges")

plt.show()
```

```python
smoker_count = data["smoker"].value_counts()

plt.pie(
    smoker_count,
    labels=smoker_count.index,
    autopct='%1.1f%%'
)

plt.title("Smoker Distribution")
plt.show()
```

```python
plt.boxplot(data["charges"])

plt.title("Medical Charges")
plt.ylabel("Charges")

plt.show()
```
