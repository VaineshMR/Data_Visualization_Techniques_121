```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("KaggleV2-May-2016.csv")

data = df.head()

plt.plot(data["Age"], range(1, len(data) + 1))

plt.title("Age vs Appointment Records")
plt.xlabel("Age")
plt.ylabel("Appointment Record")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("KaggleV2-May-2016.csv")

data = df.head()

plt.bar(data["Age"], range(1, len(data) + 1))

plt.title("Age vs Appointment Records")
plt.xlabel("Age")
plt.ylabel("Appointment Record")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("KaggleV2-May-2016.csv")

data = df.head()

status_count = data["No-show"].value_counts()

plt.pie(
    status_count,
    labels=status_count.index,
    autopct='%1.1f%%'
)

plt.title("Appointment Status Distribution")
plt.show()
```
